```python
#calculadora inteligente
print("Bem-vindo a Calculadora Inteligente!")
print("Escolha a operação que deseja realizar:")
print("1 - Soma (+)")
print("2 - Subtração (-)")
print("3 - Multiplicação (*)")
print("4 - Divisão (/)")

#usuario escolhe qual operação ele quer fazer
escolha_usuario = input("Digite o número da operação desejada:")

#vendo se o usuario fez uma escolha válida
if escolha_usuario in ("1", "2", "3", "4"):
  num1 = float(input("Digite o primeiro número:"))
  num2 = float(input("Digite o segundo número:"))
  if escolha_usuario == "1":
    resultado = num1 + num2
    print("O resultado da soma é: ", resultado)
  elif escolha_usuario == "2":
    resultado = num1 - num2
    print("O resultado da subtração é: ", resultado)
  elif escolha_usuario == "3":
    resultado = num1 * num2
    print("O resultado da multiplicação é: ", resultado)
  elif escolha_usuario == "4":
   # impede que o usuario divida por zero
    if num2 != 0:
      resultado = num1 / num2
      print("O resultado da divisão é: ", resultado)
    else:
      print("Não é possível dividir por zero. Tente outro número")

else:
  print("Opção inválida! Tente novamente.")







```
